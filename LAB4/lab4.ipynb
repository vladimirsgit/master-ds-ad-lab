```python
from pyod.utils.data import generate_data
from pyod.models.ocsvm import OCSVM
from sklearn.metrics import roc_auc_score, balanced_accuracy_score
from pyod.utils.utility import standardizer
import matplotlib.pyplot as plt
from pyod.models.deep_svdd import DeepSVDD
```

```python
# Ex 1
X_train, X_test, y_train, y_test = generate_data(n_features=3, n_train=300, n_test=200, contamination=0.15)
X_train, X_test = standardizer(X_train, X_test)
```

```python
ocsvm_model = OCSVM(contamination=0.15, kernel='linear')
ocsvm_model.fit(X_train)
test_preds = ocsvm_model.predict(X_test)
test_scores = ocsvm_model.decision_function(X_test)
train_preds = ocsvm_model.predict(X_train)
```

```python
bac = balanced_accuracy_score(y_test, test_preds)
rocauc = roc_auc_score(y_test, test_scores)
print(bac, rocauc)
```

```python
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(221, projection='3d')
ax.scatter3D(X_train[:, 0], X_train[:, 1], X_train[:, 2], c=y_train, edgecolor='black', s=100)
ax.set_title('Training set ground truth')

ax = fig.add_subplot(222, projection='3d')
ax.scatter3D(X_train[:, 0], X_train[:, 1], X_train[:, 2], c=train_preds, edgecolor='black', s=100)
ax.set_title('Training set predicted')

ax = fig.add_subplot(223, projection='3d')
ax.scatter3D(X_test[:, 0], X_test[:, 1], X_test[:, 2], c=y_test, edgecolor='black', s=100)
ax.set_title('Test set ground truth')

ax = fig.add_subplot(224, projection='3d')
ax.scatter3D(X_test[:, 0], X_test[:, 1], X_test[:, 2], c=test_preds, edgecolor='black', s=100)
ax.set_title('Test set preds')

fig.suptitle("OCSVM Linear kernel")
plt.show()
```

```python
# Ex 2:
from scipy.io import loadmat
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import make_scorer, balanced_accuracy_score
import numpy as np
```

```python
cardio =  loadmat('cardio 1.mat')
X, y = cardio['X'], cardio['y']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.6, random_state=42)

y_train = np.array([1 if val == 0 else -1 for val in y_train.flatten()])
y_test = np.array([1 if val == 0 else -1 for val in y_test.flatten()])

outlier_cnt = len(np.where(y_train == -1)[0])
cont_rate = outlier_cnt / len(y_train)
print(cont_rate)
```

```python
parameters = {
    'oc-svm__kernel' : ['rbf', 'poly', 'sigmoid'],
    'oc-svm__gamma' : [0.01, 0.1, 1, 'auto'],
    'oc-svm__nu' : [0.01, 0.05, cont_rate, 0.15, 0.2, 0.3, 0.35, 0.4, 0.45, 0.5]
}

pipeline = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('oc-svm', OneClassSVM())
])

search = GridSearchCV(estimator=pipeline, 
                      param_grid=parameters,
                      scoring=make_scorer(balanced_accuracy_score))
search.fit(X_train, y_train)
```

```python
best_model = search.best_estimator_
y_preds = best_model.predict(X_test)
bac = balanced_accuracy_score(y_test, y_preds)
print(search.best_params_)
print(bac)
```

```python
# Ex 3:
shuttle =  loadmat('shuttle 1.mat')
X, y = shuttle['X'], shuttle['y']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
X_train, X_test = standardizer(X_train, X_test)
print(X_train.shape)

outlier_cnt = len(np.where(y_train == 1)[0])
cont_rate = outlier_cnt / len(y_train)
print(cont_rate)
```

```python
ocsvm = OCSVM(contamination=cont_rate)
deepsvdd = DeepSVDD(n_features=9, preprocessing=False, contamination=cont_rate, epochs=10)

ocsvm.fit(X_train)
print("Fitted ocsvm")
deepsvdd.fit(X_train)
print("Fitted deepsvdd")

ocsvm_preds = ocsvm.predict(X_test)
ocsvm_scores = ocsvm.decision_function(X_test)
print("Predicted ocsvm")
deepsvdd_preds = deepsvdd.predict(X_test)
deepsvdd_scores = deepsvdd.decision_function(X_test)
print("Predicted deepsvdd")
ocsvm_bac = balanced_accuracy_score(y_test, ocsvm_preds)
ocsvm_rocauc = roc_auc_score(y_test, ocsvm_scores)

deepsvdd_bac = balanced_accuracy_score(y_test, deepsvdd_preds)
deepsvdd_rocauc = roc_auc_score(y_test, deepsvdd_scores)
```

```python
print("OCSVM-BAC", ocsvm_bac)
print("OCSVM-ROCAUC", ocsvm_rocauc)

print("DEEPSVDD-BAC", deepsvdd_bac)
print("DEEPSVDD-ROCAUC", deepsvdd_rocauc)
```

```python
possible_layers = [[128, 64, 16], [128, 64, 32, 16, 8], [32, 16, 8], [16, 8], [8, 4]]
for pl in possible_layers:
    dropout_value = np.random.randint(low=1, high=4) / 10
    print("\n\nTrying architecture", pl, "with dropout", dropout_value)
    deepsvdd = DeepSVDD(n_features=9, preprocessing=False, contamination=cont_rate, epochs=20, dropout_rate=dropout_value, hidden_neurons=pl, batch_size=64, random_state=42)

    deepsvdd.fit(X_train)
    print("Fitted")

    deepsvdd_preds = deepsvdd.predict(X_test)
    deepsvdd_scores = deepsvdd.decision_function(X_test)
    print("Predicted")

    deepsvdd_bac = balanced_accuracy_score(y_test, deepsvdd_preds)
    deepsvdd_rocauc = roc_auc_score(y_test, deepsvdd_scores)
    print("-- BAC", deepsvdd_bac)
    print("-- ROCAUC", deepsvdd_rocauc)
```
